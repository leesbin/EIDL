```python
import numpy as np
import os
import nazca as nd
import nazca.interconnects as IC
import nazca.demofab

################################################### design parameter #########################
resolution = 50  # 해상도
Mask_thick = 25
design_region_x = round(0.4, 2)
design_region_y = round(7.0 + 2 * Mask_thick / resolution, 2)
design_region_z = round(7.0 + 2 * Mask_thick / resolution, 2)
design_region_resolution = int(resolution)
Nx = int(design_region_resolution * design_region_x) + 1
Ny = int(design_region_resolution * design_region_y) + 1
Nz = int(design_region_resolution * design_region_z) + 1
##############################################################################################

# Numpy 배열 생성 2D 배열을 만듭니다
structure_weight = np.loadtxt('lastdesign.txt')
structure_weight = structure_weight.reshape(Nx, Ny * Nz)[Nx - 1]
data = np.rot90(structure_weight.reshape(Ny, Nz))  # ZxY

def grating(x1,y1,x2,y2,layers = 1):
    return nd.Polygon(layer=layers,
                      points = [(x1, y1),
                                (x2, y1),
                                (x2, y2),
                                (x1, y2)])

MUX_iteration = np.loadtxt('lastdesign.txt')
MUX_iteration = MUX_iteration.reshape(Nx, Ny * Nz)[Nx - 1]
data = np.rot90(structure_weight.reshape(Ny, Nz))

x=0
y=0

for j in range(Ny*Nz):
    if MUX_iteration[j] > 0.5:
        grating(1/200, 1/200, -1/200, -1/200).put(x+4+(-j//Ny)/100, y+4-(j%Nz)/100)

nd.export_gds(filename = 'chihyen.gds')
```
